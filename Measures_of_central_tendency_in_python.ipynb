```python
data = [7,8,1,9,6,4,5,3,2,1,22]

def median(data):
    n = len(data)

    # Bubble Sort
    for i in range(n):
        for j in range(0, n - i - 1):
            if data[j] > data[j + 1]:
                temp = data[j]
                data[j] = data[j + 1]
                data[j + 1] = temp

    print("Sorted data:", data)

    # Find median
    if n % 2 == 1:
        # Odd number of elements
        med = data[n // 2]
    else:
        # Even number of elements
        med = (data[n // 2 - 1] + data[n // 2]) / 2

    return med

print("Median:", median(data))
```
